```python
#Modificar un valor de la lista en un indice especifico
Alumnos[1]= "Maria Luisa"
print(Alumnos)
```

```python
# el metodo appende agrega un elemnto al final de la lista
Alumnos.append ("karla")
print(Alumnos)
```

```python
#el metodo insert agrega un elemto a la lista  EN LA POSICION ESPECIFICADA

Alumnos.insert(2,"cristina")
print(Alumnos)
```

```python
# remove elimina un elemnto especificando el valor del elemnto
#debe coincidir exactamente mayusculas y minisculus
Alumnos.remove("Ana")
```

```python
#pop elimina el elemnto o item en la posicion especificada
Alumnos.pop(2)
print(Alumnos)
```
